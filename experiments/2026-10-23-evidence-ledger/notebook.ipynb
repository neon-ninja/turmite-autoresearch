# What does the boundary-access evidence establish?
This editorial notebook reads the claim catalogue, not raw simulations. No random sampling or new estimates: random seed is not applicable. The linked studies retain their own seeds and controls. The question is which observations survive later corrections, and what decision they support.

Run from the repository root or this experiment directory under the locked uv environment. Sources are pinned; catalogue SHA-256 below identifies the exact synthesis. The catalogue is human-authored and is not a substitute for reading experiment corrections.

```python
import hashlib, json
from pathlib import Path
root = Path.cwd().resolve()
while not (root / 'Cargo.toml').exists():
    if root == root.parent:
        raise RuntimeError('Run inside the turmite repository')
    root = root.parent
path = root / 'experiments/2026-10-23-evidence-ledger/results/claims.json'
raw = path.read_bytes()
catalogue = json.loads(raw)
print('Evidence revision:', catalogue['source_revision'])
print('Catalogue SHA-256:', hashlib.sha256(raw).hexdigest())
print('Editorial records:', len(catalogue['claims']))
```

```python
for claim in catalogue['claims']:
    print(claim['id'], claim['kind'], claim['verdict'])
    print('  Next:', claim['next_decision'])
    print('  Limit:', claim['limitations'])
```

The records preserve calibration and instrument failures separately from physical rejections. They do not derive an exponent, establish a universal class, or prove sparse gates absent.
