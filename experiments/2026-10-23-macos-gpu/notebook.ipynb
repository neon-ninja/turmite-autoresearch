## Assumptions and predictions

One invocation owns one ant and each allocated tile. The ant is serial; only independent jobs run in parallel. Output must agree with the CPU oracle regardless of lane count, job order, slice length or memory pressure. No speedup factor is claimed.

## Method

The shader uses sparse 64-by-64 byte tiles, four cells per 32-bit word, paired 32-bit lifetime counters and exact SplitMix64 arithmetic for backgrounds. The CPU reconstructs the final trail and runs the existing highway detector. Host tests compare every RunResult field, with an independent seeded dense reference.

Controls include zero budgets, partial heading words, all supported colour counts, seeded high bits, tile reuse, forced pool retries, known highways and escape on the final budgeted step. ## Limits and next work

Only native census sweeps are ported. The CUDA probe and browser simulator are unchanged. Correctness does not establish throughput: dispatch, state readback, byte packing, paired arithmetic and host detection can limit gains. Run a controlled, repeated performance study before claiming a Mac speedup.
